```python
from sklearn.model_selection import train_test_split, RandomizedSearchCV, GridSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler
# Define your DataFrame and target variable
df = pd.read_csv('train_dataset.csv')  # Replace with your actual data file
y = df['winnerId']  # Define your target variable
X = df.drop(columns=['winnerId'])  # Drop the target variable from features
# Ensure X is a DataFrame
if not isinstance(X, pd.DataFrame):
    X = pd.DataFrame(X)
# Define text features
text_features = ['candidateATranscript', 'candidateBTranscript', 
                 'candidateAResume', 'candidateBResume', 'role']
# Check if all text features are in the DataFrame
missing_features = [feature for feature in text_features if feature not in X.columns]
if missing_features:
    raise ValueError(f"Missing columns in DataFrame: {', '.join(missing_features)}")
# Create a ColumnTransformer to apply TfidfVectorizer to text features
text_transformer = ColumnTransformer(
    transformers=[
        (feature, TfidfVectorizer(), feature) for feature in text_features
    ],
    remainder='drop'  # Drop other columns that are not text features
)
# Encode target variable
le = LabelEncoder()
y_encoded = le.fit_transform(y)  # Ensure y is the full target column
# Split dataset into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.2, random_state=0
)
# Define parameter grids for Random Search and Grid Search
param_dist_rf = {
    'classifier__n_estimators': [100, 200, 300, 500],
    'classifier__max_features': ['auto', 'sqrt', 'log2', 0.5],
    'classifier__max_depth': [None, 10, 20, 30, 40],
    'classifier__min_samples_split': [2, 5, 10, 15],
    'classifier__min_samples_leaf': [1, 2, 4, 6],
    'classifier__bootstrap': [True, False],
    'classifier__class_weight': [None, 'balanced']
}
param_dist_mlp = {
    'classifier__hidden_layer_sizes': [(50,), (100,), (150,), (50, 50), (100, 50)],
    'classifier__activation': ['relu', 'tanh', 'logistic'],
    'classifier__solver': ['adam', 'sgd', 'lbfgs'],
    'classifier__alpha': [0.0001, 0.001, 0.01, 0.1, 1.0],
    'classifier__learning_rate': ['constant', 'invscaling', 'adaptive']
}
# Randomized Search for RandomForestClassifier
pipeline_rf = Pipeline(steps=[
    ('preprocessor', text_transformer),
    ('classifier', RandomForestClassifier())
])

random_search_rf = RandomizedSearchCV(
    pipeline_rf,
    param_distributions=param_dist_rf,
    n_iter=50,  # Number of random searches
    cv=2,  # Number of cross-validation folds (adjust as needed)
    verbose=1,
    random_state=42,
    n_jobs=-1
)

random_search_rf.fit(X_train, y_train)

print("Random Search Best Parameters for RandomForestClassifier:")
print(random_search_rf.best_params_)
print("Best Score:", random_search_rf.best_score_)
# Grid Search for RandomForestClassifier
grid_search_rf = GridSearchCV(
    pipeline_rf,
    param_grid=param_dist_rf,
    cv=2,  # Number of cross-validation folds (adjust as needed)
    verbose=1,
    n_jobs=-1
)

grid_search_rf.fit(X_train, y_train)

print("Grid Search Best Parameters for RandomForestClassifier:")
print(grid_search_rf.best_params_)
print("Best Score:", grid_search_rf.best_score_)
# Randomized Search for MLPClassifier
pipeline_mlp = Pipeline(steps=[
    ('preprocessor', ColumnTransformer(
        transformers=[
            (feature, TfidfVectorizer(), feature) for feature in text_features
        ],
        remainder='drop'
    )),
    ('scaler', StandardScaler(with_mean=False)),  # Add scaling step
    ('classifier', MLPClassifier())
])

random_search_mlp = RandomizedSearchCV(
    pipeline_mlp,
    param_distributions=param_dist_mlp,
    n_iter=100,  # Number of random searches
    cv=2,  # Number of cross-validation folds (adjust as needed)
    verbose=1,
    random_state=42,
    n_jobs=-1
)

random_search_mlp.fit(X_train, y_train)

print("Random Search Best Parameters for MLPClassifier:")
print(random_search_mlp.best_params_)
print("Best Score:", random_search_mlp.best_score_)
# Grid Search for MLPClassifier
grid_search_mlp = GridSearchCV(
    pipeline_mlp,
    param_grid=param_dist_mlp,
    cv=2,  # Number of cross-validation folds (adjust as needed)
    verbose=1,
    n_jobs=-1
)

grid_search_mlp.fit(X_train, y_train)

print("Grid Search Best Parameters for MLPClassifier:")
print(grid_search_mlp.best_params_)
print("Best Score:", grid_search_mlp.best_score_)
```
